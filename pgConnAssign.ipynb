```python
# 10. Connection to the PostgreSQL database
import psycopg2

connection = psycopg2.connect(
    host="localhost",
    database="implementation",
    user="postgres",
    password="3609"
)

curr = connection.cursor()
```

```python
# 11. Insert with parameters

# Insert a new student using cur.execute() and %s placeholders. Do not construct the SQL using an f-string
curr.execute("INSERT INTO students (name, age, dept) VALUES (%s, %s, %s)", ("Alice", 20, "CS"))
```

```python
# 12 . Read Data

curr.execute("SELECT * FROM students")
students = curr.fetchall()
for student in students:
    print(f"ID: {student[0]}, Name: {student[1]}, Age: {student[2]}, Dept: {student[3]}")
```

```python
# 13. 13. Fetch One
curr.execute("SELECT name, marks FROM students ORDER BY marks DESC LIMIT 1")
student = curr.fetchone()
print(f"Highest Scoring Student: {student[0]}, Marks: {student[1]}")
```

```python
# 14. Update + Commit 

curr.execute("UPDATE students SET marks = marks + 5 WHERE name = %s", ("Ravi",))
connection.commit()
```

```python
# 15. Join using Pandas

import pandas as pd

df = pd.read_sql("SELECT s.name, c.course_name FROM students s JOIN enrollments e ON s.id = e.student_id JOIN courses c ON e.course_id = c.id", connection)
print(df)
```

```python
# 16. Subquery
# Use a subquery to display students whose marks are above the average marks of all students.
curr.execute("SELECT * FROM students WHERE marks > (SELECT AVG(marks) FROM students)")
students = curr.fetchall()
for student in students:
    print(f"ID: {student[0]}, Name: {student[1]}, Age: {student[2]}, Dept: {student[3]}, Marks: {student[4]}")
```

```python
# 17 . Reusable Function

def add_student(name, marks):
    curr.execute("INSERT INTO students (name, marks) VALUES (%s, %s) RETURNING id", (name, marks))
    student_id = curr.fetchone()[0]
    connection.commit()
    return student_id
```

```python
# 18. Error Handling
try:
    curr.execute("INSERT INTO students (name, marks) VALUES (%s, %s)", ("Bob", 85))
    connection.commit()
except Exception as e:
    print(f"Error occurred: {e}")
    connection.rollback()
```
